### Exercise B2 (2 points)

## Part E — Saving and Exporting Visualizations

### Exercise E1 (1 point)
Take any one plot from this notebook and save it as:
1. A PNG file at 300 dpi (`my_plot.png`)
2. An SVG vector file (`my_plot.svg`)

Briefly state (in a markdown cell) which format you'd choose for a slide presentation vs. for a scientific journal figure, and why.


````python
iris = sns.load_dataset('iris')

plt.figure(figsize=(8, 5))
sns.scatterplot(data=iris, x='sepal_width', y='sepal_length', hue='species')
plt.title('Sepal Width vs Sepal Length')
plt.savefig('my_plot.png', dpi=300, bbox_inches='tight')
plt.savefig('my_plot.svg', bbox_inches='tight')
plt.show()
````

#### Notes
For a **slide presentation**, a PNG at a reasonable dpi (150-300) is usually sufficient since slides are viewed at a fixed size on screens/projectors, and PNG is universally supported and lightweight. For a **scientific journal figure**, an SVG (or other vector format like PDF/EPS) is preferable because it can be resized/rescaled without any loss of quality — important since journals often resize figures to fit columns and print at high resolution.
